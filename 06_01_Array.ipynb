## 6.1 Introduction to Arrays

An **array** is a collection of elements of the same type stored under a **single variable name**. Each element is accessed by an **index** that starts at **0**. In Python, arrays are implemented as **lists**.

| Term | Definition | Example |
|---|---|---|
| **Array** | Collection of elements | `[10, 20, 30, 40, 50]` |
| **Element** | A single value in the array | `30` |
| **Index** | Position number (0-based) | Index 2 → value `30` |
| **Size** | Total number of elements | `len(arr)` = 5 |

Arrays let you manage many values (e.g., 500 sensor readings) with a single variable and process all of them with one loop — essential for engineering data analysis.


````python
# Worked example — stress readings at 6 points along a beam
stress_MPa = [250, 310, 285, 295, 320, 275]   # MPa

# Access by index
print(f"First reading  (index 0): {stress_MPa[0]} MPa")
print(f"Third reading  (index 2): {stress_MPa[2]} MPa")
print(f"Last reading   (index -1): {stress_MPa[-1]} MPa")
print(f"Array length: {len(stress_MPa)} readings")

# Iterate over all elements
print("All readings:")
for i, s in enumerate(stress_MPa):
    print(f"  Point {i}: {s} MPa")
````

### Exercise 6.1.3 — Material Properties Lookup (Applied)

Material data for four common engineering materials:

```python

yield_MPa   = [250, 270, 880, 200]
```

1. Print each material name alongside its yield strength.
2. Find and print the index and name of the **strongest** material (highest yield strength) without using `max()`.


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

materials  = ["Steel", "Aluminum", "Titanium", "Cast Iron"]
yield_MPa  = [250, 270, 880, 200]

# Print all materials and strengths
print("Material yield strengths:")
for i in range(len(materials)):
    print(f"  {materials[i]:<12}: {yield_MPa[i]} MPa")

# Find the strongest without max()
strongest_idx = 0
for i in range(1, len(yield_MPa)):
    if yield_MPa[i] > yield_MPa[strongest_idx]:
        strongest_idx = i

print(f"Strongest material: {materials[strongest_idx]} ({yield_MPa[strongest_idx]} MPa) at index {strongest_idx}")

# ===== SOLUTION END =====
````

## 6.2 Declaration and Initialization of Arrays

There are three common ways to create an array (list) in Python:

**Method 1 — Explicit values:**
```python
forces = [120.5, 115.2, 130.8, 118.4, 125.0]
```

**Method 2 — Default fill (all zeros, all `None`, etc.):**
```python
pressures = [0] * 10          # 10 zeros
```

**Method 3 — Using `range()` or list comprehension:**
```python
time_steps = list(range(1, 101))      # [1, 2, ..., 100]
zeros      = [0 for i in range(5)]    # [0, 0, 0, 0, 0]
```

You can also **slice** an existing array to extract a sub-array: `arr[start:stop]` (stops before `stop`).


````python
# Worked example — initialize and populate a pressure array
n = 10
pressures_bar = [0.0] * n          # start with n zeros

# Simulate filling with measured values
measured = [98.5, 99.2, 101.0, 100.5, 99.8, 100.3, 98.9, 99.6, 100.1, 99.4]
for i in range(n):
    pressures_bar[i] = measured[i]

print(f"Pressure array : {pressures_bar}")
print(f"Length         : {len(pressures_bar)}")

# Slice: extract mid-section (indices 3–6)
mid_section = pressures_bar[3:7]
print(f"Mid-section    : {mid_section}")
````

### Exercise 6.2.2 — Time Steps and Displacement Arrays (Medium)

Create two arrays using `range()` and list comprehensions:

1. `time_steps` — 11 values from 0 to 10 (inclusive), representing seconds.
2. `displacement_mm` — each value is `time_step * 2.5` mm (constant velocity motion).

Print both arrays side by side in a table: `t=0 s  →  0.0 mm`


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

time_steps      = list(range(11))                      # [0, 1, 2, ..., 10]
displacement_mm = [t * 2.5 for t in time_steps]        # [0.0, 2.5, 5.0, ...]

print(f"{'Time (s)':<12} {'Displacement (mm)'}")
print("-" * 30)
for t, d in zip(time_steps, displacement_mm):
    print(f"t={t:<5} s     {d:.1f} mm")

# ===== SOLUTION END =====
````

### Exercise 6.2.3 — Slicing a Deflection Dataset (Applied)

Deflection measurements (mm) at 10 equally-spaced points along a beam:
`deflections_mm = [0.0, 0.5, 1.2, 2.1, 3.0, 3.5, 2.8, 1.5, 0.8, 0.0]`

Use **slicing** to extract and print:
1. The first 3 readings (near-support zone).
2. The last 3 readings (far-support zone).
3. Readings at indices 3–6 (the mid-span region, 4 values).


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

deflections_mm = [0.0, 0.5, 1.2, 2.1, 3.0, 3.5, 2.8, 1.5, 0.8, 0.0]

near_support = deflections_mm[0:3]     # indices 0, 1, 2
far_support  = deflections_mm[-3:]     # last 3
mid_span     = deflections_mm[3:7]     # indices 3, 4, 5, 6

print(f"Near-support zone (idx 0–2): {near_support}")
print(f"Far-support zone  (last 3) : {far_support}")
print(f"Mid-span region   (idx 3–6): {mid_span}")

# ===== SOLUTION END =====
````
