## 6.1 Introduction to Arrays

An **array** is a collection of elements of the same type stored under a **single variable name**. Each element is accessed by an **index** that starts at **0**. In Python, arrays are implemented as **lists**.

| Term | Definition | Example |
|---|---|---|
| **Array** | Collection of elements | `[10, 20, 30, 40, 50]` |
| **Element** | A single value in the array | `30` |
| **Index** | Position number (0-based) | Index 2 → value `30` |
| **Size** | Total number of elements | `len(arr)` = 5 |

Arrays let you manage many values (e.g., 500 sensor readings) with a single variable and process all of them with one loop — essential for engineering data analysis.


````
# Worked example — stress readings at 6 points along a beam
stress_MPa = [250, 310, 285, 295, 320, 275]   # MPa

# Access by index
print(f"First reading  (index 0): {stress_MPa[0]} MPa")
print(f"Third reading  (index 2): {stress_MPa[2]} MPa")
print(f"Last reading   (index -1): {stress_MPa[-1]} MPa")
print(f"Array length: {len(stress_MPa)} readings")

# Iterate over all elements
print("
All readings:")
for i, s in enumerate(stress_MPa):
    print(f"  Point {i}: {s} MPa")
````

## 6.2 Declaration and Initialization of Arrays

There are three common ways to create an array (list) in Python:

**Method 1 — Explicit values:**
```python
forces = [120.5, 115.2, 130.8, 118.4, 125.0]
```

**Method 2 — Default fill (all zeros, all `None`, etc.):**
```python
pressures = [0] * 10          # 10 zeros
```

**Method 3 — Using `range()` or list comprehension:**
```python
time_steps = list(range(1, 101))      # [1, 2, ..., 100]
zeros      = [0 for i in range(5)]    # [0, 0, 0, 0, 0]
```

You can also **slice** an existing array to extract a sub-array: `arr[start:stop]` (stops before `stop`).


````
# Worked example — initialize and populate a pressure array
n = 10
pressures_bar = [0.0] * n          # start with n zeros

# Simulate filling with measured values
measured = [98.5, 99.2, 101.0, 100.5, 99.8, 100.3, 98.9, 99.6, 100.1, 99.4]
for i in range(n):
    pressures_bar[i] = measured[i]

print(f"Pressure array : {pressures_bar}")
print(f"Length         : {len(pressures_bar)}")

# Slice: extract mid-section (indices 3–6)
mid_section = pressures_bar[3:7]
print(f"Mid-section    : {mid_section}")
````

## 6.4 Operations on Arrays: Sum, Average, Min, Max

Four fundamental statistical operations that Python provides as built-in functions:

| Operation | Built-in | Manual loop approach |
|---|---|---|
| **Sum** | `sum(arr)` | Accumulate with `total += x` |
| **Average** | `sum(arr) / len(arr)` | `total / count` after loop |
| **Minimum** | `min(arr)` | Track `min_val`, update if smaller |
| **Maximum** | `max(arr)` | Track `max_val`, update if larger |

Always verify results make physical sense — units, realistic ranges, expected order of magnitude.


````
# Worked example — full analysis of beam deflection data
deflections_mm = [0.0, 0.5, 1.2, 2.1, 3.0, 3.5, 2.8, 1.5, 0.8, 0.0]

total_defl = sum(deflections_mm)
avg_defl   = total_defl / len(deflections_mm)
max_defl   = max(deflections_mm)
min_defl   = min(deflections_mm)
peak_idx   = deflections_mm.index(max_defl)

print(f"Number of measurement points: {len(deflections_mm)}")
print(f"Sum of deflections : {total_defl:.1f} mm")
print(f"Average deflection : {avg_defl:.2f} mm")
print(f"Maximum deflection : {max_defl:.1f} mm at point {peak_idx}")
print(f"Minimum deflection : {min_defl:.1f} mm")
print(f"Deflection range   : {max_defl - min_defl:.1f} mm")
````

### Exercise 6.4.3 — Manual Average Using a Loop (Medium)

Using the same force data as 6.4.1: `force_N = [450, 510, 480, 525, 495, 505]`

Compute the average **manually** using a `for` loop and an accumulator — do **not** use Python's `sum()`. Compare your result to the `sum()` / `len()` result to verify.


````
# TODO: Write your code here
````

````
# ===== SOLUTION START =====

force_N = [450, 510, 480, 525, 495, 505]

# Manual approach using a for loop
total_manual = 0
for f in force_N:
    total_manual = total_manual + f

average_manual = total_manual / len(force_N)

# Built-in approach for verification
average_builtin = sum(force_N) / len(force_N)

print(f"Manual average  : {average_manual:.2f} N")
print(f"Built-in average: {average_builtin:.2f} N")
print(f"Results match   : {abs(average_manual - average_builtin) < 1e-9}")

# ===== SOLUTION END =====
````

### Exercise 6.4.4 — Fatigue Test Data Analysis (Applied)

Stress amplitudes (MPa) recorded during a fatigue test:

The material's endurance limit is **180 MPa**. Compute and print:
1. The number of readings **above** the endurance limit.
2. The average stress amplitude.
3. The peak and minimum stress amplitude.
4. The **index** of the peak reading.


````
# TODO: Write your code here
````

````
# ===== SOLUTION START =====

stress_MPa     = [150, 175, 200, 175, 160, 190, 205, 180, 170, 195]
endurance_limit = 180  # MPa

# 1. Count readings above the endurance limit
above_limit = 0
for s in stress_MPa:
    if s > endurance_limit:
        above_limit += 1

# 2–4. Statistical summary
avg_stress = sum(stress_MPa) / len(stress_MPa)
peak       = max(stress_MPa)
minimum    = min(stress_MPa)
peak_index = stress_MPa.index(peak)

print(f"Total readings            : {len(stress_MPa)}")
print(f"Above endurance limit     : {above_limit} readings")
print(f"Average stress amplitude  : {avg_stress:.1f} MPa")
print(f"Peak stress               : {peak} MPa (at index {peak_index})")
print(f"Minimum stress            : {minimum} MPa")

# ===== SOLUTION END =====
````

## Challenge Problem — Multi-Beam Deflection Analysis

A structural engineer tests three beams, each measured at **5 points** along its span (mm):

```python
beam_data = [
    [0.0, 1.2, 3.5, 2.1, 0.0],   # Beam 0
    [0.0, 0.8, 2.8, 1.5, 0.0],   # Beam 1
    [0.0, 1.5, 4.2, 2.8, 0.0],   # Beam 2
]
```

Using nested loops and array operations, compute and print a summary report that includes:
1. The **maximum deflection** for each beam.
2. The **average deflection** for each beam (including end points).
3. Which beam number has the **largest maximum deflection**.
4. The overall maximum across all beams.


````
# TODO: Write your code here
````

````
# ===== SOLUTION START =====

beam_data = [
    [0.0, 1.2, 3.5, 2.1, 0.0],
    [0.0, 0.8, 2.8, 1.5, 0.0],
    [0.0, 1.5, 4.2, 2.8, 0.0],
]

max_deflections = []
avg_deflections = []

print("=" * 45)
print(f"{'Beam':<8} {'Max (mm)':<14} {'Average (mm)':<14}")
print("-" * 45)

for beam_idx, beam in enumerate(beam_data):
    # Max deflection
    max_d = beam[0]
    for d in beam:
        if d > max_d:
            max_d = d

    # Average deflection
    total = 0
    for d in beam:
        total += d
    avg_d = total / len(beam)

    max_deflections.append(max_d)
    avg_deflections.append(avg_d)
    print(f"Beam {beam_idx:<3} {max_d:<14.1f} {avg_d:<14.2f}")

print("=" * 45)

# Beam with largest max
worst_beam = 0
for i in range(1, len(max_deflections)):
    if max_deflections[i] > max_deflections[worst_beam]:
        worst_beam = i

overall_max = max_deflections[worst_beam]
print(f"
Largest max deflection: Beam {worst_beam} ({overall_max:.1f} mm)")
print(f"Overall maximum across all beams: {overall_max:.1f} mm")

# ===== SOLUTION END =====
````
