## 3.1 Selection Control Structure

A **selection control structure** tells the program to execute one block of code *or* another, depending on whether a condition is `True` or `False`.  Without it, a program would do exactly the same thing every time it runs, regardless of any sensor reading or input value.

Before writing Python code, engineers often sketch the logic as **pseudocode** — an informal, step-by-step description using plain words — or as a **flowchart** using decision diamonds (◇).  Both tools help you think through the branches before you worry about syntax.

The example below shows how a pseudocode description of a motor temperature check maps directly to Python `if`/`else`.

```python
# Worked example — motor overheat check
# Pseudocode:
#   INPUT temperature
#   IF temperature > 90 THEN
#       PRINT "Activate cooling fan"
#   ELSE
#       PRINT "System normal"
#   END IF

temperature_c = 95.0   # degrees Celsius — read from sensor

if temperature_c > 90:             # decision: is temperature above limit?
    print("Activate cooling fan")  # True branch
else:
    print("System normal")         # False branch
```

## 3.2 Boolean Data Type

The **Boolean data type** (`bool`) can hold exactly two values: `True` or `False`.  Every condition inside an `if` statement must ultimately evaluate to one of these two values.  In Python, `True` and `False` are also equivalent to the integers `1` and `0`, which is useful when you need a numeric flag (for example, a fault indicator in a data log).

A **Boolean expression** is any combination of variables, values, and operators that produces `True` or `False` — for example, `temperature > 90` or `pressure == 200`.

```python
# Worked example — Boolean values and expressions

temperature_c = 95.0
pressure_bar  = 210.0

# Boolean literals
sensor_ok = True
fault     = False
print(type(sensor_ok))    # <class 'bool'>
print(int(sensor_ok))     # 1  — True converts to 1
print(int(fault))         # 0  — False converts to 0

# Boolean expressions
print(temperature_c > 90)     # True  (95.0 > 90)
print(pressure_bar == 200)    # False (210.0 ≠ 200)
print(temperature_c <= 100)   # True  (95.0 ≤ 100)
```
