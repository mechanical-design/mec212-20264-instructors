### Exercise 2.2.3 (Independent)
Given candidate names: ["load_1", "2stress", "for", "beamLength", "yield_strength"],
print which names are valid Python identifiers and not keywords.

```python
# TODO: Check identifier validity using isidentifier and keyword checks
# Your code here
```

```python
# ===== SOLUTION START =====
import keyword

candidates = ["load_1", "2stress", "for", "beamLength", "yield_strength"]
for name in candidates:
    is_valid = name.isidentifier() and not keyword.iskeyword(name)
    print(f"{name}: {is_valid}")
# ===== SOLUTION END =====
```

## 2.3 Basic Data Types in Python
Data types define the kind of value stored: integer, decimal, text, or true/false.
Mechanical example: part count (int), diameter (float), sensor label (str), safety flag (bool).

```python
# Tutorial Example 1: int and float
bolt_count = 12
shaft_diameter_mm = 19.5

print(type(bolt_count), bolt_count)
print(type(shaft_diameter_mm), shaft_diameter_mm)
```

```python
# Tutorial Example 2: strings for labels and reports
sensor_id = "TC-07"
unit = "degC"
message = f"Sensor {sensor_id} reading unit: {unit}"
print(message)
```

```python
# Tutorial Example 3: bool from a safety comparison
stress_mpa = 180
yield_strength_mpa = 250
is_safe = stress_mpa < yield_strength_mpa

print(f"Safe condition? {is_safe}")
```

### Exercise 2.3.3 (Independent)
Given rpm_text = "1500" and max_rpm = 1800,
convert rpm_text to int, compute angular speed in rad/s, and create a bool within_limit.

```python
# TODO: Convert RPM text to numeric values and evaluate operating limit
# Your code here
```

```python
# ===== SOLUTION START =====
rpm_text = "1500"
max_rpm = 1800

rpm = int(rpm_text)
omega = 2 * 3.14159 * rpm / 60
within_limit = rpm <= max_rpm

print(f"RPM: {rpm}")
print(f"Angular speed: {omega:.2f} rad/s")
print(f"Within limit? {within_limit}")
# ===== SOLUTION END =====
```
