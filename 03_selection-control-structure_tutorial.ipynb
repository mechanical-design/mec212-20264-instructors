## 3.1 Selection Control Structure

A **selection control structure** tells the program to execute one block of code *or* another, depending on whether a condition is `True` or `False`. Without it, a program would behave identically every run, regardless of any sensor reading or input.

Before writing Python code, engineers often sketch the logic as **pseudocode** — an informal step-by-step description using plain words — or as a **flowchart** using decision diamonds (◇). Both help you think through the branches before worrying about syntax.

The example below shows how a pseudocode motor temperature check maps directly to Python `if`/`else`.

````python
# Worked example — motor overheat check
# Pseudocode:
#   INPUT temperature
#   IF temperature > 90 THEN
#       PRINT "Activate cooling fan"
#   ELSE
#       PRINT "System normal"
#   END IF

temperature_c = 95.0   # degrees Celsius

if temperature_c > 90:             # decision node
    print("Activate cooling fan")  # True branch
else:
    print("System normal")         # False branch
````

## 3.2 Boolean Data Type

The **Boolean data type** (`bool`) holds exactly two values: `True` or `False`. Every condition inside an `if` statement must ultimately evaluate to one of these. In Python, `True` and `False` are equivalent to integers `1` and `0`, which is useful when storing a numeric fault flag in a data log.

A **Boolean expression** is any combination of variables, values, and operators that produces `True` or `False` — for example, `temperature > 90` or `pressure == 200`.

````python
# Worked example — Boolean values and expressions

temperature_c = 95.0
pressure_bar  = 210.0

# Boolean literals
sensor_ok = True
fault     = False
print(type(sensor_ok))    # <class 'bool'>
print(int(sensor_ok))     # 1
print(int(fault))         # 0

# Boolean expressions
print(temperature_c > 90)     # True
print(pressure_bar == 200)    # False
print(temperature_c <= 100)   # True
````

## 3.3 Relational Operators

**Relational operators** compare two values and always return `True` or `False`. They are the basic building blocks of conditions.

| Operator | Meaning               | Example (`temp = 95`) | Result  |
|----------|-----------------------|-----------------------|---------|
| `==`     | Equal to              | `temp == 95`          | `True`  |
| `!=`     | Not equal to          | `temp != 95`          | `False` |
| `>`      | Greater than          | `temp > 90`           | `True`  |
| `<`      | Less than             | `temp < 90`           | `False` |
| `>=`     | Greater than or equal | `temp >= 95`          | `True`  |
| `<=`     | Less than or equal    | `temp <= 100`         | `True`  |

> **Common mistake:** Do not use `=` inside a condition — that is assignment. Use `==` to compare.

````python
# Worked example — relational operators on engineering values

temperature_c = 95.0
pressure_bar  = 210.0
rpm           = 1500

print(temperature_c > 90)     # True  — over-temperature?
print(pressure_bar == 200)    # False — exact match
print(rpm >= 1000)            # True  — at or above minimum speed?
print(rpm != 0)               # True  — motor running?
print(temperature_c <= 100)   # True  — still below boiling?
````

## 3.3 Logical Operators

**Logical operators** combine two or more Boolean expressions into a single `True`/`False` result, allowing you to check multiple engineering conditions at once.

| Operator | Rule                                  | Engineering use                              |
|----------|---------------------------------------|----------------------------------------------|
| `and`    | Both conditions must be `True`        | All safety interlocks must be satisfied      |
| `or`     | At least one condition must be `True` | Any fault triggers an alarm                  |
| `not`    | Inverts a Boolean value               | Is the system **not** in emergency stop?     |

**Truth table:**

| A | B | A and B | A or B | not A |
|---|---|---------|--------|-------|
| T | T | T       | T      | F     |
| T | F | F       | T      | F     |
| F | T | F       | T      | T     |
| F | F | F       | F      | T     |

````python
# Worked example — compound conditions on a pump system

temp_c         = 95.0
pressure_bar   = 280.0
emergency_stop = False

# 'and': both must be True
if temp_c > 90 and pressure_bar > 250:
    print("Critical: reduce load immediately")

# 'or': at least one must be True
if temp_c > 120 or pressure_bar > 300:
    print("Emergency: shutdown now")
else:
    print("Operating within emergency limits")

# 'not': invert a flag
if not emergency_stop:
    print("System running — no E-stop active")
````

## 3.3 Operator Precedence

When an expression contains multiple operators, Python evaluates them in a fixed order called **operator precedence**. Higher-precedence operators are evaluated first:

1. **Arithmetic:** `**`, then `*` `/`, then `+` `-`
2. **Relational:** `>`, `<`, `>=`, `<=`, `==`, `!=`
3. **`not`**
4. **`and`**
5. **`or`**

So `2 + 3 > 4 and 10 - 2 < 9` evaluates as `(2+3 > 4) and (10-2 < 9)` → `True and True` → `True`.
Use parentheses `()` to make intent explicit and avoid logic bugs.

````python
# Worked example — step-by-step precedence

load_n    = 450
area_mm2  = 150
YIELD_MPA = 3.0   # simplified value for easy arithmetic

# Arithmetic runs first, then relational
stress_exceeds = load_n / area_mm2 > YIELD_MPA
#              = (450 / 150)       > 3.0
#              = 3.0               > 3.0  →  False
print(stress_exceeds)   # False

# Compound example
result = 2 + 3 > 4 and 10 - 2 < 9
# (5 > 4) and (8 < 9) → True and True → True
print(result)   # True
````

## 3.4 One-Way Selection (`if`)

A **one-way selection** runs a block of code only when the condition is `True`. If the condition is `False`, the block is skipped entirely — nothing happens.

```python

```

Use a one-way `if` whenever you need to act on a single special case (e.g. trigger a warning) and do nothing otherwise.

````python
# Worked example — low coolant warning

coolant_level_pct = 18.0
LOW_COOLANT_PCT   = 20.0

if coolant_level_pct < LOW_COOLANT_PCT:
    print("Warning: coolant level low — refill before next shift")

print("Engine inspection complete")   # always runs
````

## 3.4 Two-Way Selection (`if–else`)

A **two-way selection** always takes exactly one of two paths. The `if` block runs when the condition is `True`; the `else` block runs when it is `False`. Unlike the one-way `if`, *something always happens*.

```python
if condition:
    statement(s)   # runs when True
else:
    statement(s)   # runs when False
```

Use `if–else` for binary engineering decisions: pass/fail, safe/unsafe, on/off.

````python
# Worked example — shaft stress pass/fail check

applied_stress_mpa = 310   # MPa — from load calculation
yield_strength_mpa = 250   # MPa — steel S275 yield strength

if applied_stress_mpa > yield_strength_mpa:
    print("FAIL: applied stress exceeds yield strength — redesign required")
else:
    print("PASS: design is safe")
````

## 3.4 Multi-Way Selection (`if–elif–else`)

A **multi-way selection** tests a series of conditions in order and executes the block belonging to the **first** condition that is `True`. All remaining conditions are then skipped. The optional `else` catches any case that did not match.

```python
if condition1:
    statement(s)    # tested first
elif condition2:
    statement(s)    # tested only if condition1 is False
elif condition3:
    statement(s)
else:
    statement(s)    # runs if nothing above matched
```

> Only the **first** matching branch runs — order conditions from most critical to least critical.

````python
# Worked example — motor RPM classification

rpm = 2200   # try: -100, 0, 500, 2200, 4000

if rpm < 0:
    print("Error: invalid RPM — cannot be negative")
elif rpm == 0:
    print("Motor stopped")
elif rpm <= 1000:
    print("Low-speed mode")
elif rpm <= 3000:
    print("Normal operating speed")
else:
    print("Warning: overspeed — check drive settings")
````

## 3.5 Nested `if` Statements

A **nested `if`** is an `if` statement placed *inside* another `if` (or `elif`/`else`) block. The inner `if` is only reached when the outer condition is `True`, creating a two-level (or deeper) decision tree.

```python
if outer_condition:          # Level 1
    if inner_condition:      # Level 2 — only reached when outer is True
        statement(s)
    else:
        statement(s)
else:
    statement(s)             # Level 1 False branch
```

Each level adds 4 spaces of indentation. More than 3 levels usually signals the logic should be restructured using `and`/`or`.

````python
# Worked example — pressure vessel safety (temperature and pressure combined)

temp_c   = 180.0
pres_bar = 35.0

if temp_c <= 200:                              # outer: temperature acceptable?
    if pres_bar <= 50:                         # inner: pressure also OK?
        print("Safe operating condition")
    else:
        print("Warning: high pressure at normal temperature")
else:                                          # outer False: temp too high
    if pres_bar <= 30:                         # reduced pressure limit
        print("Warning: high temperature — reduced pressure limit active")
    else:
        print("DANGER: both temperature and pressure exceed limits")
````

## Challenge Problem — Hydraulic Press Operating State

A hydraulic press has three sensor readings:

| Variable            | Value  | Description                     |
|---------------------|--------|---------------------------------|
| `pressure_bar`      | 210    | System pressure (bar)            |
| `oil_temp_c`        | 72     | Hydraulic oil temperature (°C)   |
| `seal_integrity_ok` | `True` | Seal passed last inspection      |

Classify the operating state by checking the rules **in order** — the first match wins:

| State    | Condition                                           | Message                                        |
|----------|-----------------------------------------------------|------------------------------------------------|
| SHUTDOWN | `pressure_bar > 250` **or** `oil_temp_c > 85`       | `"SHUTDOWN: critical parameter exceeded"`      |
| WARNING  | `pressure_bar > 200` **or** `oil_temp_c > 70`       | `"WARNING: monitor closely and reduce load"`   |
| CAUTION  | seal integrity is **not** OK                        | `"CAUTION: seal inspection overdue"`           |
| SAFE     | all other cases                                     | `"SAFE: normal operating condition"`           |

After printing the state, also print a one-line summary showing all three sensor values.

> **Hint:** This combines relational operators, logical operators, `not`, and a multi-way `if–elif–else`.

````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

pressure_bar      = 210
oil_temp_c        = 72
seal_integrity_ok = True

if pressure_bar > 250 or oil_temp_c > 85:
    state = "SHUTDOWN: critical parameter exceeded"
elif pressure_bar > 200 or oil_temp_c > 70:
    state = "WARNING: monitor closely and reduce load"
elif not seal_integrity_ok:
    state = "CAUTION: seal inspection overdue"
else:
    state = "SAFE: normal operating condition"

print(state)
print(f"Sensors — Pressure: {pressure_bar} bar | "
      f"Oil temp: {oil_temp_c} °C | "
      f"Seal OK: {seal_integrity_ok}")

# ===== SOLUTION END =====
````
