## 4.2 The `while` Loop

The **`while` loop** is a **condition-controlled** loop: it repeats as long as a Boolean condition stays `True`. Use it when the number of iterations is **not known in advance** — for example, iterating until a numerical method converges or until a sensor threshold is reached.

```
while condition:
    statement(s)    # body repeats while condition is True
```

> Always ensure the loop body changes a variable that affects the condition — otherwise you get an **infinite loop**.


````python
# Worked example — Newton-Raphson iteration to find sqrt(2)
# Iterates x_{n+1} = x_n - (x_n^2 - 2) / (2*x_n) until error < 1e-6

x   = 1.0       # initial guess
tol = 1e-6      # convergence tolerance
itr = 0         # iteration counter

while abs(x**2 - 2) > tol:      # keep going until error is small enough
    x   = x - (x**2 - 2) / (2 * x)
    itr += 1

print(f"Approximation : {x:.8f}")
print(f"True value    : {2**0.5:.8f}")
print(f"Iterations    : {itr}")
````

````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

temp_C = 900.0   # initial temperature (°C)
cooling_rate = 0.08  # fraction lost per minute
minute = 0

print(f"Minute {minute:>3}: {temp_C:.1f} °C")

while temp_C >= 100:
    temp_C = temp_C * (1 - cooling_rate)   # lose 8% each minute
    minute += 1
    print(f"Minute {minute:>3}: {temp_C:.1f} °C")

print(f"
Reached below 100°C after {minute} minutes.")

# ===== SOLUTION END =====
````

### Exercise 4.2.7 — Iterative Square Root Approximation (Applied)

Use the Babylonian method (Newton-Raphson for square root) to find √144:

$$x_{n+1} = \frac{x_n + \frac{\text{target}}{x_n}}{2}$$

Start with `x = 10.0` and `target = 144.0`. Iterate until `|x² - target| < 1e-8`. Print the result, the exact value, and the number of iterations needed.


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

target = 144.0   # the number whose square root we want
x = 10.0         # initial guess
tol = 1e-8       # convergence tolerance
itr = 0

while abs(x**2 - target) > tol:
    x = (x + target / x) / 2    # Babylonian update
    itr += 1

print(f"Computed sqrt({target:.0f}) : {x:.10f}")
print(f"Exact value            : {target**0.5:.10f}")
print(f"Iterations             : {itr}")

# ===== SOLUTION END =====
````

## 4.2 Do-While Loop (Emulated in Python)

A **do-while loop** runs its body **at least once**, then checks the condition. Python has no native `do-while` keyword, but the pattern is emulated with `while True` and `break`:

```python
while True:
    statement(s)        # body always runs at least once
    if not condition:
        break           # exit when condition becomes False
```

This is useful for **input validation** (ask at least once, repeat if invalid) or **menu-driven programs** that must display the menu before accepting a choice.


````python
# Worked example — validate that pipe diameter is positive
# Simulate sequential inputs from a list (avoids blocking input() calls)
inputs = [-5, 0, 25]    # simulated user entries
index = 0

while True:
    diameter_mm = inputs[index]       # get next simulated value
    index += 1
    print(f"Trying: {diameter_mm} mm")

    if diameter_mm > 0:               # condition check AFTER the action
        break
    print("Invalid: diameter must be > 0. Try again.")

print(f"Accepted diameter: {diameter_mm} mm")
````

### Exercise 4.2.9 — Quality Control: Accept First Good Part (Applied)

Parts leave a production line and are measured one by one. The acceptable shaft diameter is `50.0 ± 0.10 mm`. Measurements: `samples_mm = [50.18, 49.87, 50.05, 49.92, 50.12]`.

Use a do-while structure to inspect parts in order and **stop as soon as** the first acceptable part is found. Print the result of each inspection and which part number was accepted.


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

samples_mm = [50.18, 49.87, 50.05, 49.92, 50.12]
nominal_mm = 50.0
tolerance  = 0.10
index = 0

while True:
    diameter_mm = samples_mm[index]
    part_number = index + 1
    deviation   = abs(diameter_mm - nominal_mm)

    if deviation <= tolerance:
        print(f"Part {part_number}: {diameter_mm:.2f} mm — ACCEPTED")
        break                    # first good part found — stop
    else:
        print(f"Part {part_number}: {diameter_mm:.2f} mm — REJECTED (deviation {deviation:.3f} mm)")
    
    index += 1

print(f"
First acceptable part: Part {part_number} ({diameter_mm:.2f} mm)")

# ===== SOLUTION END =====
````

## 4.3 Nested Loops

A **nested loop** is a loop placed inside another loop. The **inner loop completes all its iterations** for each single iteration of the outer loop:

- Total iterations = outer count × inner count
- The outer loop typically controls rows / major steps
- The inner loop typically controls columns / minor steps

Nested loops are commonly used in engineering to process **two-dimensional data** such as sensor grids, matrices, or parameter sweeps.


````python
# Worked example — average temperature across a 3×4 sensor grid (°C)
grid = [
    [72.1, 74.3, 75.0, 73.8],   # row 0
    [80.2, 81.5, 79.9, 82.0],   # row 1
    [68.4, 67.9, 70.1, 69.5],   # row 2
]

total = 0.0
count = 0

for row_index, row in enumerate(grid):      # outer: each row
    for col_index, temp in enumerate(row):  # inner: each sensor in the row
        total += temp
        count += 1

print(f"Total readings : {count}")
print(f"Grid average   : {total / count:.2f} °C")
````

A 3×4 grid records pressure (kPa) from sensors across a manifold:
```python
pressure_grid = [
    [101.3, 102.1, 100.8, 101.5],
    [ 99.5, 100.2,  99.8, 100.0],
    [103.1, 102.5, 103.8, 102.9],
]
```
Use nested loops to compute and print:
1. The average pressure for each row.
2. The overall average across all rows.


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

pressure_grid = [
    [101.3, 102.1, 100.8, 101.5],
    [ 99.5, 100.2,  99.8, 100.0],
    [103.1, 102.5, 103.8, 102.9],
]

grand_total = 0.0
grand_count = 0

for row_idx, row in enumerate(pressure_grid):
    row_total = 0.0
    for p in row:               # inner loop: sum each row
        row_total += p
    row_avg = row_total / len(row)
    print(f"Row {row_idx} average: {row_avg:.2f} kPa")
    grand_total += row_total
    grand_count += len(row)

print(f"
Overall average: {grand_total / grand_count:.2f} kPa")

# ===== SOLUTION END =====
````

## Challenge Problem — Batch Hardness Analysis

A quality engineer measures the Brinell hardness (HB) of steel parts from **3 production batches**, each with **5 samples**:

```python
batch_data = [
    [185, 192, 178, 195, 188],   # Batch 0
    [210, 205, 212, 208, 215],   # Batch 1
    [175, 180, 172, 178, 182],   # Batch 2
]
```

Using nested loops, compute and print:
1. The average hardness for each batch.
2. The overall average across all batches.
3. Which batch number has the **highest** average hardness.

Acceptable range: 180–220 HB. Also print whether each batch average is within this range.


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

batch_data = [
    [185, 192, 178, 195, 188],
    [210, 205, 212, 208, 215],
    [175, 180, 172, 178, 182],
]

lower_limit = 180  # HB
upper_limit = 220  # HB

batch_averages = []

# Compute average for each batch using nested loops
for batch_idx, batch in enumerate(batch_data):
    batch_total = 0
    for hb in batch:
        batch_total += hb
    batch_avg = batch_total / len(batch)
    batch_averages.append(batch_avg)

    status = "PASS" if lower_limit <= batch_avg <= upper_limit else "FAIL"
    print(f"Batch {batch_idx}: average = {batch_avg:.1f} HB — {status}")

# Overall average
grand_total = 0
grand_count = 0
for batch in batch_data:
    for hb in batch:
        grand_total += hb
        grand_count += 1
overall_avg = grand_total / grand_count
print(f"
Overall average: {overall_avg:.1f} HB")

# Batch with highest average
best_batch = 0
for i in range(1, len(batch_averages)):
    if batch_averages[i] > batch_averages[best_batch]:
        best_batch = i
print(f"Highest average hardness: Batch {best_batch} ({batch_averages[best_batch]:.1f} HB)")

# ===== SOLUTION END =====
````
