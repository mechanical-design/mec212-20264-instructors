## 5.3 Parameters, Return Values, and Scope

**Default parameters** give a parameter a fallback value if no argument is provided:
```python
def weight(mass_kg, g=9.81):
    return mass_kg * g
```

A function can **return multiple values** as a tuple:
```python
def circle_properties(r):
    area = 3.14159 * r**2
    circumference = 2 * 3.14159 * r
    return area, circumference
```

**Scope** determines where a variable exists. Variables created **inside** a function are **local** — they do not exist outside it. Variables at the top level are **global**. Best practice: pass data through parameters and return values rather than using global variables.


````python
# Worked example — safety factor with validation
def safety_factor(applied_stress_MPa, yield_strength_MPa):
    sf = yield_strength_MPa / applied_stress_MPa
    return sf

# Test for steel and cast iron
materials = [
    ("Steel A36",  200, 250),
    ("Cast Iron",  180, 200),
    ("Aluminum 6061", 240, 270),
]

print(f"{'Material':<18} {'Applied (MPa)':<16} {'Yield (MPa)':<14} {'SF':<8} {'Status'}")
print("-" * 70)

for name, applied, yield_str in materials:
    sf = safety_factor(applied, yield_str)
    status = "OK" if sf >= 1.5 else "REVIEW"
    print(f"{name:<18} {applied:<16} {yield_str:<14} {sf:<8.2f} {status}")
````

### Exercise 5.3.2 — Circle Properties: Multiple Return Values

Define `circle_properties(radius_m)` that returns **both** the area (m²) and circumference (m) of a circle as a tuple. Use `pi = 3.14159`.

Call it for radius = 0.05 m and radius = 0.12 m. Unpack and print both values for each radius.


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

def circle_properties(radius_m):
    pi = 3.14159
    area_m2 = pi * radius_m**2
    circumference_m = 2 * pi * radius_m
    return area_m2, circumference_m       # return as tuple

# Call and unpack for two radii
for r in [0.05, 0.12]:
    area, circ = circle_properties(r)
    print(f"Radius = {r:.2f} m  →  Area = {area:.6f} m²,  Circumference = {circ:.4f} m")

# ===== SOLUTION END =====
````

### Exercise 5.3.3 — Safety Factor Checker with Default Threshold (Applied)

Define `safety_factor_check(applied_MPa, yield_MPa, min_sf=1.5)` that:
1. Computes the safety factor: SF = yield / applied.
2. Returns a tuple `(sf, result)` where `result` is `"PASS"` if SF ≥ min_sf, else `"FAIL"`.

Test it for these designs:

| Material | Applied (MPa) | Yield (MPa) | Min SF |
|---|---|---|---|
| Steel | 180 | 250 | 1.5 (default) |
| Aluminum | 240 | 270 | 1.5 (default) |
| High-strength bolt | 300 | 350 | 1.2 (custom) |


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

def safety_factor_check(applied_MPa, yield_MPa, min_sf=1.5):
    sf = yield_MPa / applied_MPa
    result = "PASS" if sf >= min_sf else "FAIL"
    return sf, result

test_cases = [
    ("Steel",               180, 250, 1.5),
    ("Aluminum",            240, 270, 1.5),
    ("High-strength bolt",  300, 350, 1.2),
]

print(f"{'Material':<22} {'SF':<8} {'Min SF':<10} {'Result'}")
print("-" * 50)

for name, applied, yld, min_sf in test_cases:
    sf, result = safety_factor_check(applied, yld, min_sf)
    print(f"{name:<22} {sf:<8.2f} {min_sf:<10.1f} {result}")

# ===== SOLUTION END =====
````

## Challenge Problem — Structural Member Analysis

Build a set of three functions for analysing a steel rod under axial load:

1. `axial_stress(force_N, area_m2)` → stress in **MPa**
2. `axial_strain(stress_MPa, E_GPa)` → strain (dimensionless)
3. `elongation_mm(strain, length_m)` → elongation in **mm**

Apply these to a steel rod:
- Applied force: 50 000 N
- Diameter: 20 mm (compute the cross-sectional area from this)
- E = 200 GPa (steel)
- Rod length: 0.5 m

Print the stress, strain, and elongation with clear labels. Also print whether the stress is below the yield strength of 250 MPa.


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

def axial_stress(force_N, area_m2):
    stress_Pa  = force_N / area_m2
    stress_MPa = stress_Pa * 1e-6
    return stress_MPa

def axial_strain(stress_MPa, E_GPa):
    strain = stress_MPa / (E_GPa * 1000)   # convert GPa to MPa
    return strain

def elongation_mm(strain, length_m):
    delta_m  = strain * length_m
    delta_mm = delta_m * 1000
    return delta_mm

# --- Steel rod properties ---
force_N    = 50000        # N
diameter_m = 0.020        # 20 mm → 0.020 m
E_GPa      = 200          # GPa
length_m   = 0.5          # m
yield_MPa  = 250          # MPa

# Cross-sectional area of circular rod
pi     = 3.14159
area_m2 = (pi / 4) * diameter_m**2

# Apply the three functions in sequence
stress  = axial_stress(force_N, area_m2)
strain  = axial_strain(stress, E_GPa)
delta   = elongation_mm(strain, length_m)

print(f"Force          : {force_N} N")
print(f"Diameter       : {diameter_m*1000:.0f} mm")
print(f"Area           : {area_m2*1e6:.2f} mm²")
print(f"Axial stress   : {stress:.2f} MPa")
print(f"Axial strain   : {strain:.6f}")
print(f"Elongation     : {delta:.4f} mm")

status = "below yield — design OK" if stress < yield_MPa else "EXCEEDS yield — redesign needed"
print(f"Stress vs yield: {stress:.2f} MPa vs {yield_MPa} MPa → {status}")

# ===== SOLUTION END =====
````
