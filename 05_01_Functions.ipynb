## 5.1 Functions: Concepts, Pseudocode, and Flowchart

A **function** is a named, reusable block of code that performs one specific task. Instead of writing the same calculation many times, you define it once and **call** it whenever needed. In engineering, functions mirror physical subsystems: a stress calculator, a unit converter, or a safety-factor checker each does one well-defined job.

Before writing code, use **pseudocode** to define the function's interface and algorithm:
```
FUNCTION function_name(parameter1, parameter2)
    algorithm steps
    RETURN result
END FUNCTION
```

Python's `def` keyword translates this directly.


````python
# Worked example — pseudocode → Python: Celsius to Fahrenheit
# Pseudocode:
#   FUNCTION celsius_to_fahrenheit(celsius)
#       SET fahrenheit = (celsius * 9/5) + 32
#       RETURN fahrenheit
#   END FUNCTION

def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

# Call the function for several temperatures
test_temps_C = [0, 100, -40, 37]
for temp_C in test_temps_C:
    temp_F = celsius_to_fahrenheit(temp_C)
    print(f"{temp_C:>5} °C  →  {temp_F:>7.2f} °F")
````

Formula: V = π × r² × h (use `3.14159` for π).

Call it for: radius = 0.05 m, height = 0.20 m. Also call it for radius = 0.10 m, height = 0.30 m.


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

# Pseudocode:
#   FUNCTION cylinder_volume(radius_m, height_m)
#       SET pi = 3.14159
#       SET volume = pi * radius_m^2 * height_m
#       RETURN volume
#   END FUNCTION

def cylinder_volume(radius_m, height_m):
    pi = 3.14159
    volume_m3 = pi * radius_m**2 * height_m
    return volume_m3

# Call 1
vol1 = cylinder_volume(0.05, 0.20)
print(f"r=0.05 m, h=0.20 m  →  Volume = {vol1:.6f} m³")

# Call 2
vol2 = cylinder_volume(0.10, 0.30)
print(f"r=0.10 m, h=0.30 m  →  Volume = {vol2:.6f} m³")

# ===== SOLUTION END =====
````

`measured_values = [9.97, 10.02, 10.06, 9.94, 10.04]`

For each, call the function and print `"PASS"` or `"FAIL"`.


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

# Pseudocode:
#   FUNCTION check_tolerance(measured, nominal, tolerance)
#       IF abs(measured - nominal) <= tolerance THEN
#           RETURN True
#       ELSE
#           RETURN False
#       END IF
#   END FUNCTION

def check_tolerance(measured, nominal, tolerance):
    return abs(measured - nominal) <= tolerance

nominal_mm  = 10.00
tolerance   = 0.05
measured_values = [9.97, 10.02, 10.06, 9.94, 10.04]

for i, val in enumerate(measured_values):
    result = check_tolerance(val, nominal_mm, tolerance)
    status = "PASS" if result else "FAIL"
    print(f"Bolt {i+1}: {val:.2f} mm — {status}")

# ===== SOLUTION END =====
````

## 5.2 Defining and Calling Programmer-Defined Functions

The `def` keyword creates a new function. The function must be **defined before it is called**. The syntax is:

```python
def function_name(parameter1, parameter2):
    statement(s)          # body — indented 4 spaces
    return value          # optional
```

- **Parameters** are the variable names listed inside the parentheses in the definition.
- **Arguments** are the actual values passed when you call the function.
- A **void function** performs an action (e.g., printing) without returning a value — omit the `return` statement.


````python
# Worked example — calculate axial strain from stress and modulus
# strain = stress_MPa / (E_GPa * 1000)   (unit conversion: GPa → MPa)

def calculate_strain(stress_MPa, E_GPa):
    strain = stress_MPa / (E_GPa * 1000)   # dimensionless
    return strain

# Call the function for three materials
print(f"{'Material':<15} {'Stress (MPa)':<15} {'E (GPa)':<10} {'Strain':<12}")
print("-" * 52)

data = [("Steel",    200, 200),
        ("Aluminum", 120,  70),
        ("Titanium", 600, 110)]

for material, stress, E in data:
    strain = calculate_strain(stress, E)
    print(f"{material:<15} {stress:<15} {E:<10} {strain:<12.6f}")
````

````python
# ===== SOLUTION START =====

def beam_moment(force_N, distance_m):
    moment_Nm = force_N * distance_m
    return moment_Nm

load_cases = [(500, 0.5), (1200, 1.0), (800, 0.75), (2000, 1.5)]

print(f"{'Force (N)':<12} {'Distance (m)':<15} {'Moment (N·m)':<15}")
print("-" * 42)

for force, distance in load_cases:
    M = beam_moment(force, distance)
    print(f"{force:<12} {distance:<15.2f} {M:<15.1f}")

# ===== SOLUTION END =====
````

### Exercise 5.2.4 — Bolt Shear Stress (Applied)

A shear force is applied to a bolt. The shear stress is:

$$\tau = \frac{F}{\frac{\pi}{4} d^2}$$

Define `bolt_shear_stress(force_N, diameter_m)` that returns shear stress in **MPa** (multiply Pa result by 1e-6).

Apply a force of **500 N** to bolts with diameters `[6, 8, 10, 12]` mm. Print each diameter (mm) and its shear stress (MPa).


````python
# TODO: Write your code here
````

````python
# ===== SOLUTION START =====

def bolt_shear_stress(force_N, diameter_m):
    pi = 3.14159
    area_m2    = (pi / 4) * diameter_m**2   # circular cross-section area
    stress_Pa  = force_N / area_m2
    stress_MPa = stress_Pa * 1e-6
    return stress_MPa

force_N       = 500
diameters_mm  = [6, 8, 10, 12]

print(f"{'Diameter (mm)':<16} {'Shear Stress (MPa)':<20}")
print("-" * 36)

for d_mm in diameters_mm:
    d_m  = d_mm / 1000           # convert mm to m
    tau  = bolt_shear_stress(force_N, d_m)
    print(f"{d_mm:<16} {tau:<20.3f}")

# ===== SOLUTION END =====
````
